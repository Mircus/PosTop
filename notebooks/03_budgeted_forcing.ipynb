# 03 - Budgeted forcing (experimental)
**Status: experimental, survey section 9.** Slice a cost-annotated forcing relation at increasing budgets. The headline result here is that budgeted *forcing* is monotone, but the *induced* cover and positivity are **not** -- this notebook shows both a case where a higher budget breaks a cover claim, and one where it makes a positivity witness appear.

```python
from postop import CostAnnotatedIncidence

cai = CostAnnotatedIncidence()

# A second, expensive-to-verify point only becomes visible at a
# higher budget -- it forces 'fever' but not 'cough'.
cai.assert_true('case1', 'flu', cost=0)
cai.assert_true('case1', 'fever', cost=0)
cai.assert_true('case1', 'cough', cost=0)
cai.assert_true('case2', 'fever', cost=50)  # only affordable from budget=50
cai.assert_true('case2', 'pneumonia', cost=50)
cai.assert_true('case2', 'infiltrates', cost=200)

budgets = [0, 50, 250]
print('Budgets considered:', budgets)
```

```python
for budget in budgets:
    system = cai.to_incidence_system_within(budget)
    print()
    print(f'Budget {budget}: points =', sorted(system.forcing.points))
    print('Observables =', sorted(system.forcing.observables))
    print('Does pneumonia cover {fever,cough}? ->', system.covers('pneumonia', {'fever', 'cough'}))
```

```python
# Non-monotonicity, made explicit: cover(fever, {cough}) at low vs high budget.
lo = cai.to_incidence_system_within(0)
hi = cai.to_incidence_system_within(50)

print('fever covers {cough} at budget=0:', lo.covers('fever', {'cough'}))
print('fever covers {cough} at budget=50:', hi.covers('fever', {'cough'}))
print('(case2 only becomes visible at budget=50, and it lacks cough --')
print(' this is exactly the non-monotonicity the survey flags in section 9.2.)')

cert = cai.budget_certificate('fever', {'cough'}, 50, kind='cover')
print()
print(cert.to_dict())
```
