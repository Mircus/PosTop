# 03 · Costs and budgets
Slice a cost-valued forcing relation at increasing budgets to see
which observables become available and which hypotheses remain consistent.


```python
from postop import CostForcing, CostPosTop

cf = CostForcing[str, str]()

cases = {
    'case1': {'flu': 0, 'fever': 0, 'cough': 0, 'aches': 0},
    'case2': {'pneumonia': 0, 'fever': 0, 'cough': 0, 'infiltrates': 200},
    'case3': {'tb': 0, 'fever': 0, 'cough': 0, 'night_sweats': 50},
}

for point, costs in cases.items():
    for obs, cost in costs.items():
        cf.set_cost(point, obs, cost)

cp = CostPosTop(cf)
budgets = [0, 50, 250]
print('Budgets considered:', budgets)
```

```python
for budget in budgets:
    pt = cp.at_budget(budget)
    print()
    print(f"Budget {budget}: points =", sorted(pt.forcing.points))
    print('Observables =', sorted(pt.forcing.observables))
    print('Does pneumonia cover {fever,cough}? ->', pt.covers('pneumonia', {'fever', 'cough'}))
    print('Positive(tb, {fever, cough})? ->', pt.positive('tb', {'fever', 'cough'}))
```

```python
pt_lo = cp.at_budget(0)
pt_hi = cp.at_budget(250)

cov_lo = pt_lo.explain_cover('pneumonia', {'fever', 'cough', 'infiltrates'})
cov_hi = pt_hi.explain_cover('pneumonia', {'fever', 'cough', 'infiltrates'})

print('Cover at budget=0:')
print(cov_lo)
print('Cover at budget=250:')
print(cov_hi)
```
