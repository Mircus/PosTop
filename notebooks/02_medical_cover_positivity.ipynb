# 02 - Medical diagnosis: cover and positivity
Ground the running patient example in code: build the forcing relation, query consistent hypotheses, and inspect witnesses/counterexamples via certificates.

```python
from pprint import pprint

from postop import IncidenceSystem, from_dict

patients = {
    'alice': ['fever', 'cough', 'fatigue'],
    'bob': ['fever', 'headache', 'fatigue', 'myalgia'],
    'carol': ['cough', 'runny_nose', 'sneezing'],
    'dave': ['fever', 'cough', 'chest_pain', 'infiltrates'],
    'eve': ['fever', 'elevated_WBC', 'fatigue'],
}

system = IncidenceSystem(from_dict(patients))
observed = {'fever', 'cough'}
print('Observed symptoms:', observed)
print('All observables:', sorted(system.forcing.observables))
```

```python
consistent = sorted(system.consistent_hypotheses(observed))
print('Consistent hypotheses:', consistent)

for hyp in ['fever', 'cough', 'infiltrates', 'headache']:
    info = system.explain(hyp, observed)
    print()
    print('Explanation for', hyp)
    pprint(info)
```

```python
cover_cert = system.explain_cover('infiltrates', observed)
print('Why infiltrates fails to cover observed:')
pprint(cover_cert.to_dict())

if cover_cert.counterexample:
    print('Counterexample patient profile:', patients[cover_cert.counterexample])
```

```python
U = {'fever', 'cough', 'infiltrates'}
V = set(patients['dave'])
compat = system.compatibility_certificate('infiltrates', U, V)
print('Compatibility witness in U:', compat.surviving_refinement)
print('Positive inside V?', system.positive(compat.surviving_refinement, V))
```

```python
# The formal-closed fixed-point check: reduction(U) == U, NOT just reduction(U) <= U
# (the latter -- contractivity -- holds for every U and proves nothing).
alice_profile = set(patients['alice'])
print('reduction(alice_profile) =', system.reduction(alice_profile))
print('Is formal closed:', system.is_formal_closed(alice_profile))
print('Is {"cough"} formal closed?', system.is_formal_closed({'cough'}))
```
