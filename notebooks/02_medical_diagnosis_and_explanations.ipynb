# 02 · Medical diagnosis and explanations
Ground the running patient example in code: build the forcing relation,
query consistent hypotheses, and inspect the detailed witnesses/counterexamples.


```python
from pprint import pprint

from postop import PosTop, from_dict

patients = {
    'alice': ['fever', 'cough', 'fatigue'],
    'bob': ['fever', 'headache', 'fatigue', 'myalgia'],
    'carol': ['cough', 'runny_nose', 'sneezing'],
    'dave': ['fever', 'cough', 'chest_pain', 'infiltrates'],
    'eve': ['fever', 'elevated_WBC', 'fatigue'],
}

pt = PosTop(from_dict(patients))
observed = {'fever', 'cough'}
print('Observed symptoms:', observed)
print('All observables:', sorted(pt.forcing.observables))
```

```python
consistent = sorted(pt.consistent_hypotheses(observed))
print('Consistent hypotheses:', consistent)

for hyp in ['fever', 'cough', 'infiltrates', 'headache']:
    info = pt.explain(hyp, observed)
    print('
Explanation for', hyp)
    pprint(info)
```

```python
cover = pt.explain_cover('infiltrates', observed)
print('Why infiltrates fails to cover observed:')
pprint(cover)

counterexample = cover['counterexample']
if counterexample:
    print('Counterexample patient profile:', patients[counterexample])
```

```python
U = {'fever', 'cough', 'infiltrates'}
V = set(patients['dave'])
witness = pt.check_compatibility('infiltrates', U, V)
print('Compatibility witness in U:', witness)
print('Positive inside V?', pt.positive(witness, V))
```
