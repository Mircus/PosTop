# 01 · Ext, Int, Hit, Sel, J
A tiny forcing relation to demonstrate the two adjunctions and the new explainability helpers.
We stick to deterministic prints so `nbconvert --execute` stays headless-friendly.


```python
from pprint import pprint

from postop import PosTop, OperatorSuite, from_dict

forcing_data = {
    'case_flu': ['flu', 'fever', 'cough', 'aches'],
    'case_covid': ['covid', 'fever', 'cough', 'fatigue'],
    'case_cold': ['cold', 'cough', 'runny_nose'],
    'case_allergy': ['allergy', 'sneezing', 'runny_nose'],
}

pt = PosTop(from_dict(forcing_data))
ops = OperatorSuite(pt)

print('Points:', sorted(pt.forcing.points))
print('Observables:', sorted(pt.forcing.observables))
```

```python
ext_fever, ext_info = ops.ext({'fever'}, trace=True)
int_subset, int_info = ops.Int({'case_flu', 'case_cold'}, trace=True)

print(ext_info)
print('Ext({fever}) =', sorted(ext_fever))
print(int_info)
print('Int({case_flu, case_cold}) =', sorted(int_subset))
```

```python
hit_block, hit_info = ops.hit({'case_flu', 'case_covid'}, trace=True)
sel_cluster, sel_info = ops.sel({'fever', 'cough', 'fatigue', 'aches'}, trace=True)

print(hit_info)
print('Hit({case_flu, case_covid}) =', sorted(hit_block))
print(sel_info)
print('Sel({fever, cough, fatigue, aches}) =', sorted(sel_cluster))
```

```python
j_set, j_info = ops.j({'fever', 'cough', 'fatigue', 'aches'}, trace=True)
print(j_info)
print('J({fever, cough, fatigue, aches}) =', sorted(j_set))

witness = pt.find_witness('fever', {'fever', 'cough', 'fatigue', 'aches'})
print('Witness confined to the cluster:', witness)
```

```python
cover_explanation = pt.explain_cover('flu', {'fever', 'cough'})
pprint(cover_explanation)

for line in ops.explain_chain('covid', {'fever', 'cough', 'fatigue'}):
    print(line)
```
