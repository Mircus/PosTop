# 01 - Forcing and the two adjunctions
A tiny forcing relation demonstrating `ext`/`box` (universal) and `diamond`/`rest` (existential), plus the explainability certificates. Deterministic prints, so `nbconvert --execute` stays headless-friendly.

```python
from pprint import pprint

from postop import IncidenceSystem, OperatorSuite, from_dict

forcing_data = {
    'case_flu': ['flu', 'fever', 'cough', 'aches'],
    'case_covid': ['covid', 'fever', 'cough', 'fatigue'],
    'case_cold': ['cold', 'cough', 'runny_nose'],
    'case_allergy': ['allergy', 'sneezing', 'runny_nose'],
}

system = IncidenceSystem(from_dict(forcing_data))
ops = OperatorSuite(system)

print('Points:', sorted(system.forcing.points))
print('Observables:', sorted(system.forcing.observables))
```

```python
ext_fever, ext_info = ops.ext({'fever'}, trace=True)
box_subset, box_info = ops.box({'case_flu', 'case_cold'}, trace=True)

print(ext_info)
print('ext({fever}) =', sorted(ext_fever))
print(box_info)
print('box({case_flu, case_cold}) =', sorted(box_subset))
```

```python
diamond_block, diamond_info = ops.diamond({'case_flu', 'case_covid'}, trace=True)
rest_cluster, rest_info = ops.rest({'fever', 'cough', 'fatigue', 'aches'}, trace=True)

print(diamond_info)
print('diamond({case_flu, case_covid}) =', sorted(diamond_block))
print(rest_info)
print('rest({fever, cough, fatigue, aches}) =', sorted(rest_cluster))
```

```python
reduced, reduced_info = ops.reduction({'fever', 'cough', 'fatigue', 'aches'}, trace=True)
print(reduced_info)
print('reduction({fever, cough, fatigue, aches}) =', sorted(reduced))

witness = system.find_witness('fever', {'fever', 'cough', 'fatigue', 'aches'})
print('Witness confined to the cluster:', witness)
```

```python
cover_cert = system.explain_cover('flu', {'fever', 'cough'})
pprint(cover_cert.to_dict())

for line in ops.explain_chain('covid', {'fever', 'cough', 'fatigue'}):
    print(line)
```
